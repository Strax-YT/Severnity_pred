```python
print("\n" + "="*50)
print("CALCULATING RESOURCES FOR ALL 856 PATIENTS")
print("="*50)

all_resources = []

for idx, patient in df.iterrows():
    severity = patient.get('drg_severity', 2)
    
    # Estimate severity if missing
    if pd.isna(severity):
        severity = 2  # Default medium
    
    resources = calculate_patient_resources(patient, severity)
    
    all_resources.append({
        'patient_id': patient.get('subject_id', idx),
        'age': patient.get('anchor_age'),
        'severity': severity,
        'total_cost': resources['total_cost'],
        'nurse_hours': resources['nurse_time_min'] / 60,
        'equipment_count': len(resources['equipment']),
        'medication_count': len(resources['medications']),
        'has_test': pd.notna(patient.get('test_name'))
    })

# Create results dataframe
results_df = pd.DataFrame(all_resources)

print("\n=== SUMMARY STATISTICS ===")
print(f"Total Patients Analyzed: {len(results_df)}")
print(f"Average Severity: {results_df['severity'].mean():.2f}")
print(f"Total Estimated Cost: ${results_df['total_cost'].sum():,.0f}")
print(f"Average Cost per Patient: ${results_df['total_cost'].mean():,.0f}")
print(f"Total Nurse Hours Needed: {results_df['nurse_hours'].sum():,.0f}")
print(f"Average Nurse Hours per Patient: {results_df['nurse_hours'].mean():.1f}")

# Save results
results_df.to_csv('patient_resource_calculations.csv', index=False)
print("\n✅ Saved results to patient_resource_calculations.csv")
```
